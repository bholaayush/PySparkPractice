```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import Window
from pyspark.sql import *
```

```python
data = [
    (1, "Ayush", "India", "2024-01-10", 500.0),
    (2, "Ravi", "India", "2024-01-12", 700.0),
    (3, "John", "USA", "2024-01-13", 300.0),
    (4, "Maria", "USA", "2024-01-14", 1200.0),
    (5, "Chen", "China", "2024-01-15", 900.0),
    (6, "Ayush", "India", "2024-02-01", 400.0),
    (7, "Ravi", "India", "2024-02-03", 100.0),
    (8, "John", "USA", "2024-02-04", 800.0),
    (9, "Maria", "USA", "2024-02-05", 200.0),
    (10, "Chen", "China", "2024-02-06", 1500.0),
]

schema = StructType([
    StructField("txn_id", IntegerType(), True),
    StructField("customer_name", StringType(), True),
    StructField("country", StringType(), True),
    StructField("txn_date", StringType(), True),
    StructField("amount", DoubleType(), True)
])

```

```python
df = spark.createDataFrame(data, schema) \
        .withColumn("txn_date", to_date("txn_date"))

```
